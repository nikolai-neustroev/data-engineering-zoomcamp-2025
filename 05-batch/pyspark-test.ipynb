```python
import pyspark
from pyspark.sql import SparkSession
```

```python
pyspark.__file__
```

- `SparkSession.builder`: Initializes the process of building a SparkSession, which is the entry point for interacting with Spark.
- `.master("local[*]")`: Specifies that Spark should run locally, using all available CPU cores (* means all cores).
- `.appName('test')`: Sets the name of the Spark application to "test".
- `.config("spark.driver.bindAddress", "127.0.0.1")`: Configures the driver to bind to the loopback address (127.0.0.1). This can help avoid network binding issues, especially when running locally.
- `.getOrCreate()`: Creates the SparkSession if it doesn't already exist, or returns the existing one.

```python
spark = SparkSession.builder \
    .master("local[*]") \
    .appName('test') \
    .config("spark.driver.bindAddress", "127.0.0.1") \
    .getOrCreate()

df = spark.read \
    .option("header", "true") \
    .csv('taxi_zone_lookup.csv')

df.show()
```

```python
df.write.parquet('zones')
```
